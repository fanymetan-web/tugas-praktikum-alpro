## PRAKTIKUM: TOKO DORAYAKI DORAEMON

```python
# Meminta input dari user berupa tiga variabel berbeda
nama_makanan = input("masukkan nama makanan:") # Meminta input nama makanan
nama_karakter = input("masukkan nama karakter:") # Meminta input nama karakter
kata_sifat = input("masukkan kata sifat:") # Mmeinta input kata sifat
```

```python
# Menggabungkan ketiga variabel menjadi satu string untuk nama toko
nama_toko = nama_makanan + " " + nama_karakter + " " + kata_sifat
print("Nama resmi toko:", nama_toko)
```

```python
# Membuat variabel untuk masing-masing modal
modal_doraemon = 555000 # Modal dari Doraemon
modal_nobita = 333000 # Modal dari Nobita
modal_shizuka = 222000 # Modal dari Shizuka
# Menghitung total modal yang terkumpul
total_modal = modal_doraemon + modal_nobita + modal_shizuka
print("Total modal yang terkumpul adalah Rp:", total_modal)
```

```python
# Membuat variabel jumlah dorayaki dan harga dorayaki
jumlah_dorayaki = 350
harga_dorayaki = 5000
# Menghitung total pendapatan Toko Dorayaki
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print("Total pendapatan Toko Dorayaki adalah Rp:", total_pendapatan)
```

```python
# Menghitung keuntungan Toko Dorayaki pada bulan pertama
keuntungan = total_pendapatan - total_modal
print("Keuntungan Toko Dorayaki adalah Rp:", keuntungan)
```
