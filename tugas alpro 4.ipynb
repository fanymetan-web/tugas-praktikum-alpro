```python
# ============== INPUT DATA ==============
usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (rupiah): "))
area_penggunaan = input("Masukkan lokasi penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ").lower()

# ============== TENTUKAN RISIKO DASAR ==============
risiko = "rendah"  # nilai awal
premi_persen = 0.05

# 1. Usia pengemudi
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = "tinggi"
    premi_persen = 0.1

# 2. Jenis kendaraan
if jenis_kendaraan == "motor":
    if risiko != "tinggi":
        risiko = "sedang"
        premi_persen = 0.07
elif jenis_kendaraan == "mobil":
    if risiko == "rendah":
        risiko = "rendah"
        premi_persen = 0.05

# 3. Area penggunaan
if area_penggunaan == "perkotaan":
    if risiko == "rendah":
        risiko = "sedang"
        premi_persen = 0.07
    elif risiko == "sedang":
        risiko = "tinggi"
        premi_persen = 0.1

# 4. Sistem pengaman
if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
        premi_persen = 0.07
    elif risiko == "sedang":
        risiko = "tinggi"
        premi_persen = 0.1

# ============== HITUNG PREMI ==============
premi_asuransi = premi_persen * nilai_kendaraan

# ============== PESAN DORAEMON ==============
pesan = "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!" if risiko == "tinggi" else "Risiko kendaraan masih aman menurut Doraemon"

# ============== TAMPILKAN HASIL ==============
print("\n" + "="*50)
print("📊 Kategori Risiko  :", risiko)
print("💰 Premi Asuransi   : Rp", premi_asuransi)
print("💬 Pesan Doraemon   :", pesan)
print("="*50)
```
